```python
import requests
import json

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "llama3:8b-instruct-q4_K_M"


def ollama_generate(prompt, max_tokens=200, temperature=0.7):
    """Stream a generation request to the locally-served quantized model."""
    payload = {
        "model": MODEL_NAME,
        "prompt": prompt,
        "stream": True,
        "options": {
            "num_predict": max_tokens,
            "temperature": temperature,
        },
    }
    tokens = []
    with requests.post(OLLAMA_URL, json=payload, stream=True) as resp:
        resp.raise_for_status()
        for line in resp.iter_lines():
            if not line:
                continue
            chunk = json.loads(line)
            if "response" in chunk:
                tokens.append(chunk["response"])
            if chunk.get("done"):
                break
    return "".join(tokens)

text = ollama_generate("Explain quantization in one paragraph.")
print(text)
```
